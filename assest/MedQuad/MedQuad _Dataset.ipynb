```python
from datasets import load_dataset
import json
import os

# Load dataset
dataset = load_dataset("LinhDuong/chatdoctor-200k")

# Create folder
os.makedirs("chatdoctor_dataset", exist_ok=True)

# Convert to list
data = dataset["train"].to_list()

# Save as json
with open("chatdoctor_dataset/chatdoctor200k.json", "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print("Dataset saved successfully")
```

```python
from datasets import load_dataset
import json

dataset = load_dataset("LinhDuong/chatdoctor-200k")

with open("chatdoctor_dataset/chatdoctor200k.jsonl", "w", encoding="utf-8") as f:
    for row in dataset["train"]:
        f.write(json.dumps(row, ensure_ascii=False) + "\n")

print("Saved as JSONL")
```

Save JSON Dataset in Separate Folder

```python
from datasets import load_dataset
import os
import json

# Create separate folder
folder_name = "chatdoctor_json_dataset"
os.makedirs(folder_name, exist_ok=True)

# Load dataset
dataset = load_dataset("LinhDuong/chatdoctor-200k")

# Convert train split to list
data = dataset["train"].to_list()

# File path inside separate folder
file_path = os.path.join(folder_name, "chatdoctor200k.json")

# Save as JSON
with open(file_path, "w", encoding="utf-8") as f:
    json.dump(data, f, indent=2, ensure_ascii=False)

print(f"Dataset saved successfully in: {file_path}")
```
