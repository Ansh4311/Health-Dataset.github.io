```python
import os
from datasets import load_dataset
import json

# Create separate folder
folder_name = "chatdoctor_1000_dataset"
os.makedirs(folder_name, exist_ok=True)

# Load dataset
dataset = load_dataset("LinhDuong/chatdoctor-200k")

# Take only first 1000 records
data_1000 = dataset["train"].select(range(1000)).to_list()

# Save as JSON
file_path = os.path.join(folder_name, "chatdoctor_1000.json")

with open(file_path, "w", encoding="utf-8") as f:
    json.dump(data_1000, f, indent=4, ensure_ascii=False)

print("Saved successfully at:", file_path)
print("Total records saved:", len(data_1000))
```
