```python
import os
import json
from datasets import load_dataset

# Create separate folder
folder_name = "medical_meadow_dataset"
os.makedirs(folder_name, exist_ok=True)

# Load dataset
dataset = load_dataset("medalpaca/medical_meadow_wikidoc")

# Take first 1000 rows for faster use
data_1000 = dataset["train"].select(range(1000))

# Convert to list
data_list = data_1000.to_list()

# Save as JSON
file_path = os.path.join(folder_name, "medical_meadow_1000.json")

with open(file_path, "w", encoding="utf-8") as f:
    json.dump(data_list, f, indent=4, ensure_ascii=False)

print("Dataset saved successfully!")
print("Location:", file_path)
```
