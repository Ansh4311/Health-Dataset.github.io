```python

!pip install -q kagglehub

import os
import shutil
import kagglehub
from pathlib import Path

BASE_DIR = "AI_Doctor_Datasets"
os.makedirs(BASE_DIR, exist_ok=True)

print(f"Main Folder Created: {BASE_DIR}")

def collect_images(dataset_path, output_dir):
    
    # Supported image extensions
    image_extensions = (
        ".png",
        ".jpg",
        ".jpeg",
        ".bmp",
        ".tiff",
        ".webp"
    )

    os.makedirs(output_dir, exist_ok=True)

    image_count = 0

    # Walk through all folders
    for root, dirs, files in os.walk(dataset_path):

        for file in files:

            if file.lower().endswith(image_extensions):

                source_path = os.path.join(root, file)

                # Create unique filename
                new_filename = f"img_{image_count}{Path(file).suffix}"

                destination_path = os.path.join(
                    output_dir,
                    new_filename
                )

                shutil.copy2(source_path, destination_path)

                image_count += 1

    print(f"\nCollected {image_count} images")
    print(f"Saved to: {output_dir}")

print("\n======================================")
print("DOWNLOADING DATASET 1")
print("Synthetic Medical Prescription OCR")
print("======================================")

dataset1_path = kagglehub.dataset_download(
    "priyanshuyav13/synthetic-medical-prescription-ocr-dataset"
)

print("Downloaded Dataset 1 Path:")
print(dataset1_path)

dataset1_output = os.path.join(
    BASE_DIR,
    "Medical_Prescription_Images"
)

collect_images(dataset1_path, dataset1_output)

print("\n======================================")
print("DOWNLOADING DATASET 2")
print("Illegible Medical Prescription Images")
print("======================================")

dataset2_path = kagglehub.dataset_download(
    "mehaksingal/illegible-medical-prescription-images-dataset"
)

print("Downloaded Dataset 2 Path:")
print(dataset2_path)

dataset2_output = os.path.join(
    BASE_DIR,
    "Illegible_Prescription_Images"
)

collect_images(dataset2_path, dataset2_output)

print("\n======================================")
print("DOWNLOAD COMPLETE")
print("======================================")

print(f"""
Datasets Ready:

1. Medical Prescription Images
   Folder:
   {dataset1_output}

2. Illegible Prescription Images
   Folder:
   {dataset2_output}
""")
```

```python
!pip install opencv-python-headless pillow imagehash scikit-learn pandas
!pip install -q opencv-python pillow imagehash scikit-learn pandas
import sys

!{sys.executable} -m pip install opencv-python-headless

import os
import cv2
import shutil
import imagehash
import pandas as pd

print(cv2.__version__)
from PIL import Image
from sklearn.model_selection import train_test_split

dataset_folders = [
    "Medical_Prescription_Images",
    "Illegible_Prescription_Images"
]

clean_folder = "Cleaned_Dataset"

train_dir = os.path.join(clean_folder, "train")
valid_dir = os.path.join(clean_folder, "valid")
test_dir  = os.path.join(clean_folder, "test")

os.makedirs(train_dir, exist_ok=True)
os.makedirs(valid_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)

image_paths = []
hashes = set()

size = (512, 512)

count = 0

for folder in dataset_folders:

    for file in os.listdir(folder):

        path = os.path.join(folder, file)

        try:
            
            img_pil = Image.open(path).convert("L")

            img_hash = imagehash.average_hash(img_pil)

            if img_hash in hashes:
                continue

            hashes.add(img_hash)

            img = cv2.imread(path)

            # Grayscale
            gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

            # Resize
            gray = cv2.resize(gray, size)

            # Denoising
            denoise = cv2.fastNlMeansDenoising(gray)

            # OCR preprocessing
            processed = cv2.adaptiveThreshold(
                denoise,
                255,
                cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                cv2.THRESH_BINARY,
                11,
                2
            )

            # Save
            filename = f"img_{count}.png"

            save_path = os.path.join(clean_folder, filename)

            cv2.imwrite(save_path, processed)

            image_paths.append(save_path)

            count += 1

        except:
            pass

print("Total Clean Images:", len(image_paths))

train, temp = train_test_split(
    image_paths,
    test_size=0.2,
    random_state=42
)

valid, test = train_test_split(
    temp,
    test_size=0.5,
    random_state=42
)

def move_files(files, folder):

    for file in files:

        shutil.copy(file, folder)

move_files(train, train_dir)
move_files(valid, valid_dir)
move_files(test, test_dir)

print("Train:", len(train))
print("Valid:", len(valid))
print("Test :", len(test))

labels = []

for img_path in image_paths:

    labels.append({
        "image_path": img_path,
        "text": ""
    })

df = pd.DataFrame(labels)

df.to_csv("ocr_labels.csv", index=False)

print("OCR CSV Created: ocr_labels.csv")
```
