```python
# Install required library (run once)
!pip install datasets pandas -q

from datasets import load_dataset
import pandas as pd
import json
import os

# Create folder
save_folder = "symptoms_diagnosis_dataset"
os.makedirs(save_folder, exist_ok=True)

# Load dataset from Hugging Face
dataset = load_dataset("sajjadhadi/disease-diagnosis-dataset")

# Take first 1000 rows from train split
df = dataset["train"].select(range(1000)).to_pandas()

# Save as JSON
json_path = os.path.join(save_folder, "symptoms_to_diagnosis_1000.json")

df.to_json(json_path, orient="records", indent=4)

print("Dataset saved successfully")
print("Location:", json_path)

# Preview first 5 rows
print(df.head())
```

Only Symptoms + Diagnosis Keys

```python
clean_data = []

for _, row in df.iterrows():
    clean_data.append({
        "symptoms": row["text"],
        "diagnosis": row["diagnosis"]
    })

clean_json_path = os.path.join(save_folder, "clean_symptoms_diagnosis_1000.json")

with open(clean_json_path, "w", encoding="utf-8") as f:
    json.dump(clean_data, f, indent=4)

print("Clean JSON saved at:", clean_json_path)
```
